```python
# Check for negative monthly charges
import pandas as pd

# Check for negative values in MonthlyCharges
negative_charges = customer_info_db[customer_info_db['MonthlyCharges'] < 0]
print(f"Number of customers with negative monthly charges: {len(negative_charges)}")

if len(negative_charges) > 0:
    print("\n" + "="*50)
    print("CUSTOMERS WITH NEGATIVE MONTHLY CHARGES")
    print("="*50)
    print(negative_charges)
    
    # Get statistics on monthly charges
    print("\n" + "="*50)
    print("MONTHLY CHARGES STATISTICS (BEFORE FIX)")
    print("="*50)
    print(customer_info_db['MonthlyCharges'].describe())
    
    # Fix negative values by converting to positive (sign error correction)
    print("\n" + "="*50)
    print("FIXING NEGATIVE VALUES")
    print("="*50)
    customer_info_db['MonthlyCharges'] = customer_info_db['MonthlyCharges'].abs()
    
    # Verify fix
    negative_after = customer_info_db[customer_info_db['MonthlyCharges'] < 0]
    print(f"Negative values after fix: {len(negative_after)}")
    
    # Show the corrected records
    corrected_customers = customer_info_db[customer_info_db['CustomerID'].isin(negative_charges['CustomerID'])]
    print("\nCorrected records:")
    print(corrected_customers[['CustomerID', 'MonthlyCharges']])
    
    # Export the cleaned data
    customer_info_db.to_csv(r"D:\Ahmed\My Projects\telecom intelligence project\data\processed\customer_info_db.csv", index=False)
    print("\nCleaned data exported successfully!")
    
    # Show statistics after fix
    print("\n" + "="*50)
    print("MONTHLY CHARGES STATISTICS (AFTER FIX)")
    print("="*50)
    print(customer_info_db['MonthlyCharges'].describe())
else:
    print("\nNo negative monthly charges found.")
    print("\nMonthly charges statistics:")
    print(customer_info_db['MonthlyCharges'].describe())
```

```python
# Check for repeated customer IDs in customer_info_db dataset
customer_info_db = pd.read_csv(r"D:\Ahmed\My Projects\telecom intelligence project\data\processed\customer_info_db.csv")

# Check for duplicate CustomerIDs
duplicate_count = customer_info_db['CustomerID'].duplicated().sum()
print(f"Number of duplicate CustomerIDs: {duplicate_count}")

# Get unique count of CustomerIDs
unique_customers = customer_info_db['CustomerID'].nunique()
total_rows = len(customer_info_db)
print(f"Total rows: {total_rows}")
print(f"Unique customer IDs: {unique_customers}")

# If duplicates exist, show them
if duplicate_count > 0:
    print("\n" + "="*50)
    print("DUPLICATE CUSTOMER IDs FOUND")
    print("="*50)
    
    # Get all duplicate rows
    duplicates = customer_info_db[customer_info_db['CustomerID'].duplicated(keep=False)]
    print(f"\nAll rows with duplicate CustomerIDs ({len(duplicates)} rows):")
    print(duplicates.sort_values('CustomerID'))
    
    # Show which customer IDs are duplicated
    duplicated_ids = customer_info_db[customer_info_db['CustomerID'].duplicated()]['CustomerID'].unique()
    print(f"\nCustomer IDs that appear more than once ({len(duplicated_ids)} unique IDs):")
    for cust_id in duplicated_ids:
        count = customer_info_db[customer_info_db['CustomerID'] == cust_id].shape[0]
        print(f"  {cust_id}: {count} occurrences")
else:
    print("\nNo duplicate CustomerIDs found in the dataset.")
```
